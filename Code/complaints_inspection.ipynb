## Importing Libraries

```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt
from sodapy import Socrata
import seaborn as sns
import os 
from dotenv import load_dotenv 
```

## Reading In Data

```python
load_dotenv() 
api = os.getenv('MY_API')
MyAppToken = os.getenv('MyAppToken')
```

```python
client = Socrata("data.cityofnewyork.us",
                  MyAppToken,
                  )
```

```python
results = client.get("erm2-nwe9", limit=1000000, where=" agency = 'NYPD' and created_date > '2022-01-01T00:00:00' and created_date < '2026-01-01T00:00:00' ")
```

## Data Inspection

## Identifying Null Values

```python
# calculating percentage of nulls values for each column
missing_data = (
    results_df
    .isnull()
    .sum()
    .div(results_df.shape[0])
    .mul(100)
)

missing_data.sort_values(ascending=False)
```

## Selection of Impactful Variables

```python
NYPD = results_df[['created_date', 'closed_date','complaint_type','resolution_description','borough','latitude','longitude', 'x_coordinate_state_plane','y_coordinate_state_plane']]
NYPD
```

```python
NYPD[ NYPD['location_type'] == 'Subway']
```

## Measuring Time to Complaint Resolution


```python
NYPD['duration'] = (NYPD['closed_date'] - NYPD['created_date'])
NYPD
```

```python
NYPD['duration in hours'] = NYPD['duration'].dt.total_seconds() / 3600
NYPD
```

## Creating Extra Variables

```python
# creating a variable where 1 indicates action taken, and 0 indicates no action taken
NYPD.loc[NYPD["resolution_description"] == "The Police Department responded to the complaint and took action to fix the condition.", "took_action"] = 1
NYPD.loc[NYPD["resolution_description"] != "The Police Department responded to the complaint and took action to fix the condition.", "took_action"] = 0

```
